### `Task` Train a KNN model on glass type dataset and find best n_neighnours.

Data Link: https://drive.google.com/file/d/17cbDNBmys04MJqQfrma3jd72VPMnxIq0/view?usp=share_link

```python
# Your code goes here
from sklearn.neighbors import KNeighborsClassifier
```

```python
from  sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,f1_score,confusion_matrix,precision_score,recall_score,classification_report
```

```python
import pandas as pd
import numpy as np
```

```python
glass_data=pd.read_csv('glass.csv')
```

```python
glass_data.iloc[:-2,:]
```

```python
x_train,x_test,y_train,y_test=train_test_split(glass_data.iloc[:,:-2],glass_data.iloc[:,-1],random_state=None,test_size=0.2)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
# Try Simple KNN Classifier with default paramerters
knn_default_parameters=KNeighborsClassifier()
```

```python
knn_default_parameters.fit(x_train,y_train)
```

```python
y_pred_with_default_parameters=knn_default_parameters.predict(x_test)
```

```python
accuracy_score(y_test,y_pred_with_default_parameters)
```

```python
confusion_matrix(y_test,y_pred_with_default_parameters)
```

```python
print(classification_report(y_test,y_pred_with_default_parameters))
```
